# IGS metadata parser
The IGS satellite metadata file as defined in [this pdf](https://files.igs.org/pub/resource/working_groups/multi_gnss/Metadata_SINEX_1.10.pdf) it provides information about different identifiers, PRN, frequency and plane allocations, satellite masses and geometries.

The parser is imported and the example file is parsed.

```python
from midgard import parsers
from pathlib import Path
path = Path.cwd().parent.parent / "tests" / "parsers" / "example_files" / "sinex_satellite_metadata"
p = parsers.parse_file(parser_name="sinex_satellite_metadata", file_path=path)
```

p.data contains nested dictionaries for each satellite (SVN is the unique identifier). Note that the data in the example metadata sinex is heavily reduced to aid clarity of function.

```python
for key, values in p.data.items():
    print(f"{key}:{values.keys()}")
```

```python
#How to find the orbital slot of "E102"
prn_E232, plane_E232 = p.data['E232']['satellite_prn'], p.data['E232']['satellite_plane']
for prn in prn_E232:
    print(prn)
for plane in plane_E232:
    print(plane)
```

Here we see that E232 has only ever had the prn of 16, this was valid from doy 261 of 2024, and is still valid.
From doy 261 in 2024 to doy 363 in 2025 it was placed in Plane 1 Slot A17
From doy 263 in 2025 to doy 032 in 2026 it was repositioning.
From doy 032 in 2026 and onwards it has been placed in Plane 1 Slot A08

A similar structure of dates is used for all metadata that may change over the course of a satellites lifetime.
